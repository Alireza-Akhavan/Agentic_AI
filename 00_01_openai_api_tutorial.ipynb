برای استفاده از OpenAI API به یک **API Key** نیاز داری. مراحل دریافت:

3. از منوی بالا سمت راست روی **API Keys** کلیک کن

5. کلید را کپی و در جای امنی ذخیره کن (بعد از بستن پنجره دیگر نمایش داده نمی‌شود)

## 🔑 مرحله ۵ — تنظیم کلید API (روش مستقیم)

در این روش کلید را مستقیماً در کد قرار می‌دهیم.

> ⚠️ این روش فقط برای تست محلی مناسب است. در پروژه‌های واقعی از متغیر محیطی استفاده کن (مرحله بعد).

````python
from openai import OpenAI

# ─── روش ۱: اتصال مستقیم به سرور OpenAI ───────────────────────────
client = OpenAI(
    api_key="sk-proj-...",   # <-- کلید خود را اینجا بگذار
)

print("✅ اتصال مستقیم به OpenAI آماده است")
````

## 🚀 مرحله ۶ — اولین درخواست

````python
# ارسال یک پیام ساده به مدل
response = client.chat.completions.create(
    model="gpt-4o-mini",   # می‌توانی مدل را عوض کنی: gpt-4o, gpt-3.5-turbo, ...
    messages=[
        {"role": "system", "content": "تو یک دستیار مفید هستی که به فارسی پاسخ می‌دهی."},
        {"role": "user",   "content": "پایتخت ایران کجاست؟"}
    ]
)

# نمایش پاسخ
print(response.choices[0].message.content)
````

### سرویس ایرانی

🔗 [https://gapgpt.app/platform-v2/](https://gapgpt.app/platform-v2/)


````python
from openai import OpenAI

# ─── روش ۲: استفاده از endpoint ایرانی GapGPT ─────────────────────
client = OpenAI(
    api_key="sk-...",   # <-- کلید خود را اینجا بگذار
    base_url="https://api.gapgpt.app/v1"
)

print("✅ اتصال از طریق endpoint ایرانی آماده است")
````

````python
# ارسال یک پیام ساده به مدل
response = client.chat.completions.create(
    model="claude-sonnet-4-6",   # می‌توانی مدل را عوض کنی: gpt-4o, gpt-3.5-turbo, ...
    messages=[
        {"role": "system", "content": "تو یک دستیار مفید هستی که به فارسی پاسخ می‌دهی."},
        {"role": "user",   "content": "پایتخت ایران کجاست؟"}
    ]
)

# نمایش پاسخ
print(response.choices[0].message.content)
````

## 📊 مرحله ۷ — بررسی مصرف توکن

````python
# اطلاعات مصرف توکن
usage = response.usage
print(f"توکن‌های ورودی  : {usage.prompt_tokens}")
print(f"توکن‌های خروجی : {usage.completion_tokens}")
print(f"مجموع توکن‌ها  : {usage.total_tokens}")
````
