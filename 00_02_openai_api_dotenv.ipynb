## 📝 مرحله ۲ — ساخت فایل `.env`

در همان پوشه‌ای که نوت‌بوک هست، یک فایل `.env` بساز.


حالا فایل `.env` را باز کن و مقدار `your-api-key-here` را با کلید واقعی‌ات جایگزین کن:

```
OPENAI_API_KEY=sk-proj-xxxxxxxxxxxxxxxxxxxxxxxx
```

اگر می‌خواهی از endpoint ایرانی GapGPT استفاده کنی، خط کامنت‌شده را هم فعال کن:

```
OPENAI_API_KEY=sk-proj-xxxxxxxxxxxxxxxxxxxxxxxx
OPENAI_BASE_URL=https://api.gapgpt.app/v1
```

## 🔒 مرحله ۳ — اضافه کردن `.env` به `.gitignore`

اگر از Git استفاده می‌کنی، این کد را اجرا کن:

````python
gitignore_path = ".gitignore"
entry = ".env"

# بررسی اینکه .env قبلاً اضافه شده یا نه
if os.path.exists(gitignore_path):
    with open(gitignore_path, "r") as f:
        content = f.read()
    if entry in content:
        print("ℹ️ .env قبلاً در .gitignore بوده")
    else:
        with open(gitignore_path, "a") as f:
            f.write(f"\n{entry}\n")
        print("✅ .env به .gitignore اضافه شد")
else:
    with open(gitignore_path, "w") as f:
        f.write(f"{entry}\n")
    print("✅ فایل .gitignore ساخته شد و .env به آن اضافه شد")
````

## 🚀 مرحله ۴ — لود کردن `.env` و اتصال به API

````python
from dotenv import load_dotenv
from openai import OpenAI
import os

# لود کردن متغیرهای محیطی از فایل .env
load_dotenv()
````

````python
client = OpenAI()
print("✅ Client آماده است")
````

## ✅ مرحله ۵ — تست اتصال

````python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "تو یک دستیار مفید هستی که به فارسی پاسخ می‌دهی."},
        {"role": "user",   "content": "سلام! یک جمله انگیزشی برایم بگو."}
    ]
)

print("پاسخ مدل:")
print(response.choices[0].message.content)
````
